Question 9: Write Python code to fit a simple linear regression model using scikit-learn and print the slope and intercept.
(Include your Python code and output in the code box below.)

```python
import numpy as np
from sklearn.linear_model import LinearRegression

# 1. Prepare the data
# X must be a 2D array (samples, features) for scikit-learn
X = np.array([1, 2, 3, 4, 5]).reshape(-1, 1)
y = np.array([45, 55, 70, 80, 92])

# 2. Initialize and fit the model
model = LinearRegression()
model.fit(X, y)

# 3. Retrieve the slope (coefficient) and intercept
slope = model.coef_[0]
intercept = model.intercept_

print(f"Slope (Beta 1): {slope:.2f}")
print(f"Intercept (Beta 0): {intercept:.2f}")

# Example Prediction
new_hour = np.array([[6]])
prediction = model.predict(new_hour)
print(f"Predicted score for 6 hours: {prediction[0]:.2f}")
```
